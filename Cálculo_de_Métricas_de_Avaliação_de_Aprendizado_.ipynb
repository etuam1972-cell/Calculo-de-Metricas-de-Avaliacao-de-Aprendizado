```python
def metricas(vp, vn, fp, fn):
    """Calcula acurácia, sensibilidade, especificidade, precisão e F-score."""
    n = vp + vn + fp + fn

    acuracia = (vp + vn) / n
    sensibilidade = vp / (vp + fn)
    especificidade = vn / (vn + fp)
    precisao = vp / (vp + fp)
    f_score = 2 * (precisao * sensibilidade) / (precisao + sensibilidade)

    return {
        "Acurácia": round(acuracia, 4),
        "Sensibilidade": round(sensibilidade, 4),
        "Especificidade": round(especificidade, 4),
        "Precisão": round(precisao, 4),
        "F-score": round(f_score, 4),
    }


# Matriz de confusão escolhida
VP, VN, FP, FN = 50, 100, 5, 10

resultado = metricas(VP, VN, FP, FN)
for nome, valor in resultado.items():
    print(f"{nome:15s}: {valor}")
```
